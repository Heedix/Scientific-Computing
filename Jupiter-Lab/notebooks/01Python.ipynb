## Leap year
A year is a leap year if it is divisible by 4. Century years are an exception: they are only leap years if they are also divisible by 400. For example, 1900 was not a leap year, while 2000 was. The example year 2024 is divisible by 4 and not by 100, so it is a leap year.

```python
year = 2024
is_leap_year = year % 400 == 0 or (year % 4 == 0 and year % 100 != 0)
print(is_leap_year)
```

## Temperatures and conversion
temps.sort() sorts the list in place. Within the loop, if and elif select the values, and append() adds them to the respective lists. The + operator concatenates the two lists. Fahrenheit values are appended to a separate list, leaving the Celsius list unchanged. The list comprehension with round(t, 2) rounds only the displayed values.

```python
temps = [25.2, 16.8, 31.4, 23.9, 28, 22.5, 19.6]
temps.sort()

cool_temps = []
warm_temps = []
for temperature in temps:
    if temperature < 20:
        cool_temps.append(temperature)
    elif temperature > 20:
        warm_temps.append(temperature)

temps_in_celsius = cool_temps + warm_temps
temps_in_fahrenheit = []
for temperature in temps_in_celsius:
    temps_in_fahrenheit.append(temperature * 9 / 5 + 32)

print("Sorted temperatures:", temps)
print("Cool:", cool_temps)
print("Warm:", warm_temps)
print("Celsius:", temps_in_celsius)
print("Fahrenheit:", [round(t, 2) for t in temps_in_fahrenheit])
```

## Sum of the first n natural numbers
range(1, n + 1) includes all integers from 1 through n because the upper bound is excluded. total += number accumulates their sum. The second function uses n * (n + 1) // 2, where // keeps the result as an integer. == compares the two return values.

```python
def sum_with_loop(n):
    total = 0
    for number in range(1, n + 1):
        total += number
    return total

def sum_with_gauss(n):
    return n * (n + 1) // 2

n = 10
loop_result = sum_with_loop(n)
gauss_result = sum_with_gauss(n)
print("Loop:", loop_result)
print("Gauss formula:", gauss_result)
print("Results agree:", loop_result == gauss_result)
```

## Factorial: recursion and loop
factorial_recursive() calls itself with n - 1 until the base case n == 0 returns 1. In factorial_loop(), result starts at 1 and is updated with *= for each value in range(2, n + 1). For n = 0 or n = 1, this range is empty, so the result remains 1.

```python
def factorial_recursive(n):
    if n == 0:
        return 1
    return n * factorial_recursive(n - 1)

def factorial_loop(n):
    result = 1
    for number in range(2, n + 1):
        result *= number
    return result

n = 5
print("Recursive:", factorial_recursive(n))
print("Loop:", factorial_loop(n))
```

## Arithmetic mean and median
sum(numbers) and len(numbers) provide the total and the number of elements for the mean. sorted() creates an ordered copy for the median. Since indexing starts at zero, the two middle values of a ten-element list are at indices 4 and 5. The function returns both results as a tuple, which is unpacked into mean and median.

```python
def mean_and_median(numbers):
    mean = sum(numbers) / len(numbers)
    ordered = sorted(numbers)
    median = (ordered[4] + ordered[5]) / 2
    return mean, median

numbers = [12, 4, 7, 9, 3, 15, 8, 6, 10, 2]
mean, median = mean_and_median(numbers)
print("Mean:", mean)
print("Median:", median)
```

## Volume and surface area of a cuboid
cuboid() takes height, width, and length as parameters and calculates both results using arithmetic operators. return volume, surface_area returns a tuple, the assignment outside the function unpacks it into two variables.

```python
def cuboid(height, width, length):
    volume = height * width * length
    surface_area = 2 * (height * width + height * length
                        + width * length)
    return volume, surface_area

volume, surface_area = cuboid(2, 3, 4)
print("Volume:", volume)
print("Surface area:", surface_area)
```

## Keyboard input and 19% VAT
input() returns a string, and strip() removes leading and trailing whitespace. Decimal() converts the input directly to a decimal value. quantize(Decimal("0.01"), rounding=ROUND_HALF_UP) rounds amounts to cents, with halfway cases rounded up for positive prices. The f-string format :.2f displays two decimal places.

```python
from decimal import Decimal, InvalidOperation, ROUND_HALF_UP

entered_price = input("Net price in euros: ").strip()
net_price = Decimal(entered_price)
net_price = net_price.quantize(Decimal("0.01"), rounding=ROUND_HALF_UP)


vat_rate = Decimal("0.19")
vat_amount = (net_price * vat_rate).quantize(
    Decimal("0.01"), rounding=ROUND_HALF_UP)
gross_price = net_price + vat_amount
print(f"Net price:   EUR {net_price:.2f}")
print(f"VAT (19%):   EUR {vat_amount:.2f}")
print(f"Gross price: EUR {gross_price:.2f}")
```

## First n Fibonacci numbers
range(n) controls the number of iterations, and append(a) stores the current value. In a, b = b, a + b, both expressions on the right are evaluated before either variable is updated. This advances the sequence without a temporary variable.

```python
def fibonacci(n):
    result = []
    a, b = 0, 1
    for index in range(n):
        result.append(a)
        a, b = b, a + b
    return result

n = 10
print(f"First {n} Fibonacci numbers:", fibonacci(n))
```
