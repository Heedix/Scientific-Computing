# Zweiter Kernel: Bash und mathematische Werkzeuge

Dieses Notebook verwendet den **Bash-Kernel**, nicht Python. Es ist für den Docker-Container vorgesehen.
Der Kernel wird in den Notebook-Metadaten vorausgewählt. Bei Bedarf oben rechts **Bash** auswählen.
Alle Zellen nacheinander ausführen. Die Programme laufen in der Linux-Umgebung des Containers.


```bash
printf 'Aktuelles Verzeichnis: '
pwd
printf '\nPython-Version: '
python --version
printf '\nInstallierte Jupyter-Kernel:\n'
jupyter kernelspec list
```

## Python und MPI aus der Shell
`mpiexec` startet vier getrennte Python-Prozesse. Der Arbeitsordner kann je nach Start des Notebooks
`/workspace` oder `/workspace/notebooks` sein; die nächste Zelle wechselt ausdrücklich zum Projektordner.


```bash
cd /workspace
mpiexec --oversubscribe -n 4 python examples/mpi_sum.py
```

## MPI mit C
`mpicc` ergänzt beim Kompilieren die für MPI notwendigen Optionen. Der Compiler erzeugt
die Binärdatei im lokal eingebundenen Ordner `examples`.


```bash
cd /workspace
mpicc -Wall -Wextra -O2 examples/mpi_hello.c -o examples/mpi_hello && \
    mpiexec --oversubscribe -n 4 examples/mpi_hello
```

## Maxima und GP
Hier reicht die Shell Standard-Eingaben an die beiden Programme weiter.


```bash
maxima --very-quiet -r 'display2d:false$ print(factor(x^4-1))$ quit();'
```

```bash
gp -q -f <<'GP'
print(factor(360))
print(isprime(101))
quit
GP
```

## PARI-C-Bibliothek
Dasselbe mathematische System kann über eine Bibliothek direkt aus C angesprochen werden.


```bash
cd /workspace
gcc -Wall -Wextra examples/pari_demo.c -lpari -o examples/pari_demo && examples/pari_demo
```
