## 2. Einführung, Überblick und Anwendungen von Scientific Computing

Scientific Computing verbindet mathematische Modelle, Algorithmen und Computerprogramme.
Anwendungen sind etwa die Simulation von Wärmeleitung, die Anpassung von Messdaten,
Optimierung sowie die Lösung großer linearer Gleichungssysteme.

Ein typischer Ablauf ist **Fragestellung → Modell → Diskretisierung → Algorithmus → Rechnung → Prüfung**.
Bei jeder Stufe können Fehler entstehen: Modellfehler, Diskretisierungsfehler und Rundungsfehler

**Numerisch:** ein Näherungswert mit endlicher Genauigkeit. **Symbolisch:** Rechnen mit exakten

````python
import math
import numpy as np
import sympy as sp
import matplotlib.pyplot as plt
from IPython.display import display

print('Gleitkomma:', 0.1 + 0.2)
print('Exakter Bruch:', sp.Rational(1, 10) + sp.Rational(2, 10))
print('Vergleich mit Toleranz:', math.isclose(0.1 + 0.2, 0.3))

# Beispiel: numerische Integration von x² auf [0, 1] mit der Mittelpunktregel.
n = 1000
midpoints = (np.arange(n) + 0.5) / n
integral = np.mean(midpoints**2)
print('Integral ≈', integral, '| absoluter Fehler:', abs(integral - 1/3))
````

## 3. Einführung in die Programmiersprache Python

Python verwendet Einrückungen für Blöcke. `**` ist Potenzieren, `*` ist Multiplikation.
Listen speichern Folgen von Objekten; Dictionaries ordnen Schlüsseln Werte zu.
Eine Funktion bündelt eine Berechnung und liefert mit `return` ihr Ergebnis zurück.


````python
name = 'Scientific Computing'
measurements = [1.0, 1.2, 0.9, 1.1]
settings = {'einheit': 'm', 'anzahl': len(measurements)}

def mean_value(values):
    if len(values) == 0:
        raise ValueError('Mindestens ein Messwert ist erforderlich.')
    return sum(values) / len(values)

print(name, settings)
print('Mittelwert:', mean_value(measurements))
print('Quadrate:', [k**2 for k in range(1, 6)])
for index, value in enumerate(measurements):
    print(index, 'über 1' if value > 1 else 'höchstens 1')
````

Python-Listen und NumPy-Arrays verhalten sich unterschiedlich. Eine Liste mit `* 2`
wird wiederholt, ein numerisches Array wird elementweise mit 2 multipliziert.
Indizes beginnen bei 0; bei `a[start:stop]` gehört der Index `stop` nicht mehr dazu.


````python
values = [1, 2, 3]
array = np.array(values, dtype=float)
print('Liste:', values * 2)
print('Array:', array * 2)
print('Erstes Element:', array[0], '| erste zwei:', array[:2])

# Fehler gezielt abfangen, wenn sie erwartet werden.
try:
    mean_value([])
except ValueError as error:
    print('Erwarteter Eingabefehler:', error)
````

## 6. Parallel Computing mit MPI

### 6.1 Überblick über den MPI-Standard
MPI (Message Passing Interface) beschreibt Kommunikation zwischen Prozessen.
Jeder Prozess hat eigenen Speicher und einen **Rang** innerhalb eines **Kommunikators**.
`COMM_WORLD` umfasst hier alle gestarteten Prozesse; die Ränge reichen von 0 bis `size - 1`.
MPI ist ein Standard, Open MPI die in diesem Projekt installierte Implementierung, `mpi4py` die Python-Anbindung.

### 6.2 Prozesskommunikation
- **send / recv:** Nachricht zwischen zwei Prozessen; Quelle, Ziel und Tag müssen zusammenpassen.
- **bcast:** denselben Wert von einem Prozess an alle verteilen.
- **gather:** lokale Ergebnisse am Wurzelprozess sammeln.
- **reduce:** Ergebnisse mit einer Operation wie Summe zusammenführen.

Alle beteiligten Prozesse müssen kollektive Aufrufe in zusammenpassender Reihenfolge ausführen.
Falsche Reihenfolgen oder fehlende Empfangsoperationen können zu Deadlocks führen.

### 6.3 Anwendungen mit Python und C
Die Beispiele liegen unter `examples/`. Wir starten eigene MPI-Prozesse über `mpiexec`;
der Notebook-Kernel wird dadurch nicht selbst parallel. `--oversubscribe` erlaubt die kleine
Lernübung auch mit wenigen verfügbaren CPU-Kernen und ist eine Open-MPI-Option.
Die Reihenfolge der Prozessausgaben ist nicht festgelegt. Mehr Prozesse bedeuten bei kleinen Aufgaben
oft mehr Laufzeit durch Start- und Kommunikationsaufwand.


````python
import shutil
import subprocess

def run_command(args, *, input_text=None, timeout=60):
    result = subprocess.run(args, input=input_text, text=True, capture_output=True,
                            cwd=PROJECT, timeout=timeout)
    if result.returncode != 0:
        raise RuntimeError(f'Befehl fehlgeschlagen: {args}\n{result.stdout}\n{result.stderr}')
    print(result.stdout.strip())
    return result.stdout

print((EXAMPLES / 'mpi_sum.py').read_text())
if shutil.which('mpiexec'):
    mpi_output = run_command(['mpiexec', '--oversubscribe', '-n', '4', sys.executable,
                              str(EXAMPLES / 'mpi_sum.py')])
    assert 'Gesamtsumme: 5050' in mpi_output
    assert 'Prozess 1 empfaengt' in mpi_output
else:
    print('Übersprungen: MPI fehlt. Für diese Übung den Docker-Container verwenden.')
````

### 8.1 Iterative Nullstellenbestimmung
Für $f(x)=x^2-2$ verwenden wir Newton:
$$x_{k+1}=x_k-\frac{f(x_k)}{f'(x_k)}.$$
Die Methode kann nahe einer einfachen Nullstelle schnell konvergieren. Sie ist aber nicht für
jeden Startwert geeignet; eine verschwindende Ableitung ist problematisch.
Wir prüfen sowohl den Schritt als auch das Residuum $|f(x_k)|$ und begrenzen die Iterationen.


````python
def newton(f, df, x0, tol=1e-12, maxiter=50):
    if tol <= 0 or maxiter < 1:
        raise ValueError('Positive Toleranz und mindestens eine Iteration erforderlich.')
    current = float(x0)
    history = [current]
    for _ in range(maxiter):
        value = f(current)
        if abs(value) <= tol:
            return current, np.array(history)
        derivative = df(current)
        if abs(derivative) < 1e-15:
            raise ValueError('Ableitung zu klein: anderen Startwert wählen.')
        updated = current - value / derivative
        if not math.isfinite(updated):
            raise ValueError('Nicht endlicher Iterationswert.')
        history.append(updated)
        if abs(updated - current) <= tol * (1 + abs(updated)) and abs(f(updated)) <= tol:
            return updated, np.array(history)
        current = updated
    raise RuntimeError('Keine Konvergenz innerhalb der maximalen Iterationszahl.')

root, history = newton(lambda z: z*z - 2, lambda z: 2*z, x0=1.0)
print('Nullstelle:', root, '| Residuum:', abs(root*root - 2))
print('Iterierte:', history)
assert np.isclose(root, np.sqrt(2), atol=1e-12, rtol=0)
````

````python
fig, ax = plt.subplots(figsize=(7, 3))
error = np.abs(history - np.sqrt(2))
ax.semilogy(range(len(history)), np.maximum(error, np.finfo(float).eps), 'o-')
ax.set(xlabel='Iteration k', ylabel='Absoluter Fehler (unten begrenzt auf ε)', title='Newton-Konvergenz')
ax.grid(alpha=0.3)
fig.tight_layout()
plt.show()
````

### 8.2 Polynom-Interpolation
Zu $n+1$ paarweise verschiedenen Stützstellen existiert genau ein Polynom vom Grad höchstens $n$,
das alle Werte trifft. Interpolation ist keine Ausgleichsrechnung bei verrauschten Daten.
Höhere Grade verbessern die Näherung nicht zwangsläufig; gleichmäßige Stützstellen können
starke Randoszillationen verursachen (Runge-Phänomen).

Wir verwenden die baryzentrische Darstellung aus SciPy, statt ein Vandermonde-System explizit zu invertieren.


````python
from scipy.interpolate import BarycentricInterpolator

nodes = np.array([-1.0, 0.0, 1.0, 2.0])
node_values = np.array([1.0, 0.0, 1.0, 4.0])
interpolant = BarycentricInterpolator(nodes, node_values)
assert np.allclose(interpolant(nodes), node_values)
print('Interpolierter Wert bei 0.5:', float(interpolant(0.5)))
display(sp.interpolate([(-1, 1), (0, 0), (1, 1), (2, 4)], x))

dense = np.linspace(-1, 2, 300)
fig, ax = plt.subplots(figsize=(7, 3))
ax.plot(dense, interpolant(dense), label='Interpolationspolynom')
ax.scatter(nodes, node_values, color='tab:red', label='Stützstellen', zorder=3)
ax.set(xlabel='x', ylabel='p(x)', title='Interpolation von vier Punkten')
ax.legend()
ax.grid(alpha=0.3)
fig.tight_layout()
plt.show()
````

### 8.3 Wichtige Operationen auf Matrizen und Vektoren
`A * B` multipliziert elementweise; `A @ B` ist das Matrixprodukt.
Für reelle Vektoren ist `u @ v` das Skalarprodukt. `np.linalg.norm` liefert die euklidische
Vektornorm; `.T` transponiert eine Matrix.


````python
A = np.array([[3.0, 1.0], [1.0, 2.0]])
u = np.array([1.0, 2.0])
v = np.array([3.0, 4.0])
print('Skalarprodukt:', u @ v)
print('Norm von u:', np.linalg.norm(u))
print('A mal u:', A @ u)
print('Elementweise A*A:\n', A * A)
print('Matrixprodukt A@A:\n', A @ A)
print('Transponierte:\n', A.T)
print('Determinante:', np.linalg.det(A))
eigenvalues, eigenvectors = np.linalg.eigh(A)  # A ist reell symmetrisch
print('Eigenwerte:', eigenvalues)
assert np.allclose(A @ eigenvectors, eigenvectors * eigenvalues)
````

### 8.4 Lineare Gleichungssysteme
Gesucht ist $x$ mit $Ax=b$. Verwende `np.linalg.solve(A, b)` für quadratische reguläre Systeme.
Das explizite Berechnen von `inv(A) @ b` ist dafür normalerweise unnötig.
Das Residuum $\|Ax-b\|$ prüft die erfüllte Gleichung. Bei schlecht konditioniertem $A$ garantiert
ein kleines Residuum allein keine kleine Abweichung von der gesuchten Lösung.
Für überbestimmte Systeme ist `np.linalg.lstsq` eine passende Ausgleichsmethode.


````python
b = np.array([9.0, 8.0])
solution = np.linalg.solve(A, b)
print('Lösung:', solution)
print('Residuum:', np.linalg.norm(A @ solution - b))
print('Konditionszahl:', np.linalg.cond(A))
assert np.allclose(A @ solution, b)
display(sp.Matrix([[3, 1], [1, 2]]).LUsolve(sp.Matrix([9, 8])))

# Überbestimmt: Gerade y = m*t + c an vier Messpunkte anpassen.
t = np.array([0.0, 1.0, 2.0, 3.0])
y = np.array([1.0, 2.9, 5.2, 6.9])
design = np.column_stack([t, np.ones_like(t)])
coefficients, residuals, rank, singular_values = np.linalg.lstsq(design, y, rcond=None)
print('Steigung m, Achsenabschnitt c:', coefficients)
print('Norm des Ausgleichsresiduums:', np.linalg.norm(design @ coefficients - y))
````

**Abschlussübung:** Löse das System
`2*x + y - z = 8`, `-3*x - y + 2*z = -11`, `-2*x + y + 2*z = -3`
mit NumPy und SymPy. Vergleiche die Ergebnisse und prüfe das Residuum.

answer = np.linalg.solve(M, rhs)
print(answer)  # [2., 3., -1.]
assert np.allclose(M @ answer, rhs)
